# Matrix multiplicaton using Functions

```python
# Initialise C to zero
# I need to consider two matrices A and B.
# I need to find the dot product of two lists.
# I need to pick ith row and jth coloumn in a matrix.
```

```python
def initialise_mat(dim):
  C=[]
  for i in range(dim):
    C.append([])
  for i in range(dim):
    for j in range (dim):
      C[i].append(0)
  return C
def dot_product(u,v):
  dim=len(u)
  ans=0
  for i in range (dim):
    ans=ans+(u[i]*v[i])
  return ans
def row(M,i):
  dim=len(M)
  l=[]
  for k in range(dim):
    l.append(M[i][k])
  return l
def coloumn(M,j):
    dim=len(M)
    l=[]
    for k in range(dim):
      l.append(M[k][j])
    return l
def matrix_mul(A,B):
  dim=len(A)
  C=initialise_mat(dim)
  for i in range (dim):
    for j in range (dim):
      C[i][j]=dot_product(row(A,i),coloumn(B,j))
  return C
A=[[1,2,3],[4,5,6],[7,8,9]]
B=[[1,2,1],[6,2,3],[4,2,1]]
print(matrix_mul(A,B))
```
