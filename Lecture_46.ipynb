# More on Dictionaries
List use square brackets, tuple use round brackets, whereas, dictionary use curly bracket.
In addition to that, every dictionary element follow a standard structure which is refered as key value pair, which has two component seprated by colon. the left side of colon is refered that key, whereas, right-side of colon is refered that value.In you want to access a particular value in a dictionary we can do so by using key associated against the value, therefore, every key in a dictionary has to be an unique value, whereas,duplication is allowed for values.

So, the next question is apart from this what all things we can use a dictionary key.
Can be use integer, as a dictionary key?
Yes, we can use integer as a dictionary key, float yes,  boolean yes, string yes,list no,because list is mutable.Another dictionary once again,no,because dictionary is also mutable


Can be used tuple as a dictionary key?
Yes, tuple is immutable hence used as dictionary
No, tuple allows to have mutable value key inside it.
Dictionary key is mutable and hashable.
Dictionary value takes anything no restriction inside are allowed as dict key.  

```python
# How we can iterate over a dictionary
d = {0:0, 1:1, 2:4,3:9,4:16,5:25}
print(d)
for key in d:
  print(key,d[key])
```

```python
d = {0:0,1:1 ,2:4,3:9,4:16,5:25}
print(d.keys())
# it print keys in this dictionary
print(d.values())
# it print values in this dictionary
print(d.items())
# but this one is bit different,
# it also prints a list, but every element in the list is a tuple
# and that couple holds two elements, key as well as value.
```
