```python
import matplotlib.pyplot as plt
import numpy as np
x=np.arange(-10,10,1)
y=3*x-4
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.arange(-10,10,1)
y=x**2+2*x-15
plt.plot(x,y)
plt.show()
```
