```python
x = 'PYTHon string method '
print(x.lower()) # all letters in lower case
print(x.upper()) # all letters in upper case
print(x.capitalize()) # first letter capital
print(x.swapcase()) # all letters changed if letter in upper then changed in lower and lower to upper.
print(x.title())   # the first letter of all word changed in upper case.
```

```python
# If you would check in bool the string is or not.
x='python'
print(x.islower())
y='Python'
print(y.islower())
```

```python
x='PYTHON'
print(x.isupper())
y='Python'
print(y.isupper())
```

```python
x='Python String Methods'
print(x.istitle())
y='1234567890'
print(y.isdigit())
z='abc'
print(z.isalpha())
```

```python
x='abc123'
print(x.isalnum())
```

```python
x='_________python_______'
print(x.strip('_'))
print(x.lstrip('_'))
print(x.rstrip('_'))
```

```python
x='python methods'
print(x.startswith('p'))
print(x.startswith('o'))
print(x.endswith('s'))
print(x.endswith('h'))
```

```python
x='python strings methods practice'
print(x.count('t'))
print(x.index('t'))
print(x.replace('r','R'))
```
