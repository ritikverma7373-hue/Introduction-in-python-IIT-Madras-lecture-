# Scope of variables

```python
def myFunction1(x):
  x=x*2 # This is local x means different version of x
  print("Value of x in a function 1",x)
def myFunction2(x):
  x=x*3 # This is the local x
  print("value of function 2",x)
x=5     # This is the global  x means only single version of x
print("Value of x before function call",x)
myFunction1(x)
myFunction2(x)
print("Value of x after function call",x)
```

This is because of two different concepts.
First, call by value and second, scope of a variable.
Because computer can maintain two different types of variables. One, which is global and second is local.

```python
def myFunction1():
  global x
  x=x*2
  print("Value of x in a function 1",x)
def myFunction2():
  global x
  x=x*3
  print("value of function 2",x)
x=5
print("Value of x before function call",x)
myFunction1()
myFunction2()
print("Value of x after function call",x)
```

So, let us see what happens with this particular keyword global. Whenever we say global x, computer realizes there is one variable name x somewhere in the global code which we have written, then it looks for that particular variable and find its value 5. And then in that particular function, it always refers to this particualar variable x.Because of that this x into 2 is 5 into 2, result is 10.
When we execute second function, once again we are saying global x, once again computer realizes that x is a global variable that means, there is only single version of this variable x.
Therefore, it will not create another copy of x which is local to this function. It will simply use the existing global copy and as the value of x is 10,it will do 10 into 3 and result into 30.
