# More on sets

sets are curly brackets, which are similiar to dictionary, whereas values inside follow a pattern, which is like a list or a tuple.
So, the question is what is different about set as compared to other three components? First, set does not allow duplicate values. if we add duplicate values and try to print the set still, we get only the unique values in that set and all the duplicates will be discarted.
Second, set is an unordered entity, which means we cannot say that this element 1 is at zeroth index. sets do not have notion of order. Set is simply a collection  of values.
Third, with respect to mutability set is peculiar. Set itself is mutable, but every value inside set has to be immutable and hashable, which means you can add values into a set,but whatever values which we add, those values has to be immutable.

Moving to next point which is set methods.The concept of set python is derived from mathmatical sets. Therefore, python set also support all those operations, which we usually perform using mathmatical sets like subset, superset,union, intersection, difference and so on.
All these mathematical operations can be implemented using python sets a specific operator or using set methods.

```python
st={1,2,3,4,5,6,1,2,3}
print(st)
```

```python
st={1,2,3,4,2,3,65,7}
print(st[0]) # We do not assign with a index
```

```python
A={1,2,3}
B={1,2,3,4,5}
print(A.issubset(B))
print(A.issuperset(B))
```

```python
# As I mentioned earlier pyhton can perform these mathematical operations
# in two different ways. First, using a method, second using a specific operator
# And this is the operator for union and intersection.
A={1,2,3}
B={3,4,5}
C1=A.union(B)
C2=A|B
print(C1,C2)
C1=A.intersection(B)
C2=A&B
print(C1,C2)
```

```python
# Difference method and difference operator.
A={1,2,3}
B={3,4,5}
C1=A.difference(B)
C2=A-B
print(C1,C2)
```
