# Tutorial on for loop and difference between while loop and for loop

In this tutorial, we will see another type of loop called for loop, for loop has its own advantages as compared to while loop, we will use same examples which we saw with while loop and try to convert them into python codes using for loop

Now, the question is how to decide that what are the factors which we should consider before deciding whether  to use while loop or for loop?
So the inportant factor we should consider is weather do we know the range of the loop are not,as in aware in advance that the loop is going to run for some n number of times, if we are certain that is execute for let us say 10 times or 12 times than we should opt for For loop.
whereas, if we do not know how many number of times the particular loop is going to execute,we should opt for while loop.

```python
# problem 1: while loop
'''Find the factorial of the given number'''
num=int(input('Enter a number: '))
fact=1
if(num<0):
  print('not defined')
else:
   while (num>0):
    fact=fact*num
    num=num-1
   print(fact)
```

```python
# problem 1: For loop
num=int(input('Enter a number: '))
fact=1
if (num<0):
   print('not defined')
else:
   for i in range (num,1,-1):
    fact=fact*i
   print(fact)

```

In this particular case, the number of iterations cannot be predicated, because the number entered by user. Hence, we cannot predict how many number of times this while loop will execute.
