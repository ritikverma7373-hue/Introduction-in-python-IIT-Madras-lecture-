# Recursion- An Illustration

```python
n=10
ans=0
for i in range (n):
  ans=ans+(i+1)
print(ans)
```

```python
def sum(n):
  ans=0
  for i in range (n):
    ans=ans+(i+1)
  return ans
print(ans)
```

# Recursion in python

```python
# recursion
def sum(n):
  if (n==1):
    return 1
  else:
    return n+sum(n-1)
print(sum(10))
```

```python
# compute the compound interest by assuming the interest to be 10 %.
def comp(p,n):
  if (n==1):
    return p*1.1
  else:
    return (comp(p,n-1))*1.1
print(comp(2000,3))
```

```python
def fact(n):
  if (n==1):
    return 1
  else:
    return fact(n-1)*n
print(fact(int(input())))
```
