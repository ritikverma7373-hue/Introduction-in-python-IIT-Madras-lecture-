# Binary search pecursionn way

```python
def rbinary_search(l,k,begin,end):
    if (begin==end) :
      if (l[begin]==k):
       return 1
      else:
       return 0
    if (end-begin==1):
      if (l[begin]==k or l[end]==k):
        return 1
      else:
        return 0
    if (end-begin>1):
       mid=(begin+end)//2
       if l[mid]>k:
        end=mid-1
       if l[mid]<k:
        begin=mid+1
       if l[mid]==k:
        return 1
    if (end-begin<0):
      return 0
    return rbinary_search(l,k,begin,end)
print(rbinary_search([1,2,10,16,100,108,1000],7,0,6))
l=list(range(1000*1000*100))
print(rbinary_search(l,1000000,0,len(l)-1))
print(rbinary_search(l,-1,0,len(l)-1))
```
