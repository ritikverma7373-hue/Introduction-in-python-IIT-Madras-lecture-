# More on tuples

 List is mutable

 Tuple is immutable

 Tuples may not allowed to modify it values.

 Tuples can access elements in the tuple using their index.

 We can perform slicing operation on tuples similiar like lists or strings

Tuple also provide methods like list except method like: append, remove, insert, pop for obivious reason, which is correct, Tuple is immutable.

We will use tuple for packing and unpacking.


```python
t=1,2,3   #packing these three values into a single entity called tuple.
print(t,type(t))
x,y,z=t   #we are unpacking this tuple into three independent integer values.
print(x,y,z)
```

```python
# Python is actually packing these two variables,y,x into a tuple
# and then unpacking it on the left-handside.
x=5
y=10
x,y=y,x
print(x,y)
```

```python
l=[10]
print(l,type(l))
t=(10)
print(t,type(t))
t=(10,)
print(t,type(t))
```

```python
# t is a tuple but the value inside tuple are two lists.
# As you see, it is possible.
t=([1,2],['a','b'])
# t[0] = [10,20] if execute we getting an error because tuples are immutable.
print(t)
```

We cannot modify tuple values, but if a value inside a tuple is a list, then we can modify the values inside that list.
In python programming language , this particular principle is refered as hashable.

```python
t=([1,2],['A','B'])
t[0][0]=10
print(t)
```

If the values inside tuple are mutable then tuple is refered as non-hasable.
Remember this term hashable,
whereas,if values inside tuple are mutable then tuple is refered as nun-hasable.

```python
t1=(1,2,3)
t2=([1,2,3])
print(t1,type(t1))
print(t2,type(t2))
```
