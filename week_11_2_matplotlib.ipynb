```python
import matplotlib.pyplot as plt
import numpy as np
x=np.arange(-7,7,0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)
y_tan=np.tan(x)
plt.plot(x,y_sin ,label='sin(x)')
plt.plot(x,y_cos ,label='cos(x)')
plt.plot(x,y_tan ,label='tan(x)')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[6,8,9,4,7,8,9,8]
y=[4,6,7,8,9,9,8,10]
plt.scatter(x,y)
plt.xlabel('Mathematics score')
plt.ylabel('Physics score')
plt.title('Mathematics vs Physics scatter plot')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
n=100
throws=np.random.randint(1,7,size=n)
probabilities=[]
for i in range (1,7):
  count=np.sum(throws==i)
  prob=count/n
  probabilities.append(prob)
plt.bar([1,2,3,4,5,6],probabilities)
plt.xlabel('Number of Dice')
plt.ylabel('Probabilities of Dice')
plt.title('pobabilities of dice outcomes')


```
