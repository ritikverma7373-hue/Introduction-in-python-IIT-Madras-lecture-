# Types of Function Arguments

Every fuctions have various different components.
The first component is called function defination which starts with keyword def.
Next is called function call, this is the place where are actually executing this defined function.
The third component is arguments, the values which we pass along with the function call are refered as function argument

```python
def add(a,b,c): # function defination keyword def
  return (a+b-c) # a,b,c are parameters
print(add(20,30,40)) # it is function call
#The values which we pass along with the function call
#are refered as function argument
```

The first parameter is assign with first argument and second assign with second argument and third parameter is assign with third argument.

```python
def add(c,a,b):
  return (a+b-c)
print(add(40,20,30))
# not everytime you will see the parameteres to assign to arguments so what's
# we do
print(add(a=20,b=30,c=40))
```

All these arguments are mapped against parameters based on its position and because of that this type of arguments called as positional argument.

number of parameter = number od argumen

```python
def add(c,a=20,b=30):
  return (a+b-c)
print(add(40))
print(add(40,10))
print(add(40,10,50))
print(add(40,b=10,a=50))
```
