# Equality operator

In a first case, it says true, and in a second case, it says false.
Which means the equality operator not only checks the values in the list, it also checks whether the sequence of these values is identical or not.

```python
l1=[1,2,3]
l2=[1,2,3]
l3=[1,3,2]
print(l1==l2)
print(l2==l3)
```

Now, you must we wondering how computer can compare two lists.This is happening because computer compares two lists using the elements one at a time.The elements at 0th index is compared against element at 0th index in second list. And so on with 1st, 2nd 3rd and so on for both two lists.

```python
print([1,2]<[2,1])
print([1]<[1,2,3])
print([2,3]<[3])
print([]<[1])
```

```python
# How lists behave when we pass list variable as an argument to a function.
# lists behave ina different manner as compared as integer.
# if the function argument is of mutable type then it is called by reference.
# Otherwise, it is called by value.
# In that case of lists, it is called by reference, whereas in case of intger,
# it was called by value.
def add(x):
  x=x+1
  return x
x=5
print(add(x))
print(x)
def add(x):
  x.append(1)
  return x
x=[5]
print(add(x))
print(x)
```
