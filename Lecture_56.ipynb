# Caesar cipher

''' This program considers on input on input file and encrypts it by using caesar cipher. By that we mean, we swift the letters by 3 units.
For example: a becomes d, b becomes e, and so on ...'''

```python
import string
def create_caesar_dictionary():
  l=string.ascii_lowercase
  l=list(l)
  d={}
  for i in range(len(l)):
    d[l[i]]=l[(i+3)%26]
  return d
  print(d)
```
