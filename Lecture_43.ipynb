### More examples on Functions

```python
#Let us try a few functions on lists.
def list_min(l):
  mini=l[0]
  for i in range (len(l)):
    if (l[i]<mini):
      mini=l[i]
  return mini
l=[7,2,3,4,5,-2,-10,9]
print(list_min(l))
```

```python
def list_max(l):
  maxi=l[0]
  for i in range (len(l)):
    if l[i]>maxi:
      maxi=l[i]
  return maxi
l=[2,3,5,7,8,9,-10,-8]
print(list_max(l))
```

```python
def appendbefore(l,z):
  newl=[]
  for i in range(len(z)):
    newl.append(z[i])
  for i in range(len(l)):
    newl.append(l[i])
  return newl
l=[1,2,3]
z=[4,5,6]
print(appendbefore(l,z))
```

```python
def appendend(l,z):
  newl=[]
  for i in range(len(l)):
    newl.append(l[i])
  for i in range(len(z)):
    newl.append(z[i])
  return newl
l=[1,2,3]
z=[-1,-2,-3]
print(appendend(l,z))
```

```python
def list_average(l):
  sum=0
  for i in range (len(l)):
    sum=sum+l[i]
  ans=sum/len(l)
  return ans
l=[1,2,3,4,5]
print(list_average(l))
```
