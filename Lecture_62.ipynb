# Sorting recursively

To find the sorted list.

```python
def mini(l):
  mini=l[0]
  for x in l:
    if (x<mini):
      mini=x
  return mini
def sort(l):
    if (len(l)==1) or (l==[]):
      return l


    m=mini(l)
    l.remove(m)
    return [m]+sort(l)
l=[1,35,23,5,53,2,1,3,41,13,31]
print(sort(l))
```
