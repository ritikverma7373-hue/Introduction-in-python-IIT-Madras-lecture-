```python
# Problem 1: write a python code to find the number of characters,  uppercase letters, lowercase letters and words.
def upper(s):
  upper = 0
  for c in s:
    if (c.isupper()):
     upper+=1
  return upper
def lower(s):
  lower=0
  for  c in s:
     if (c.islower()):
      lower+=1
  return lower
def characters(s):
  characters=0
  for c in s:
    characters+=1
  return characters
def words(s):
  words=1
  for c in s:
    if c==' ':
      words+=1
  return words
sentence=input("Enter a sentence: ")
uletters=upper(sentence)
lletters=lower(sentence)
chars=characters(sentence)
wds=words(sentence)
print(f'/n the number of upper case letters is {uletters}')
print(f'/n the number of lower case letters is {lletters}')
print(f'/n the number of characters is {chars}')
print(f'/n the number of words is {wds}')


```
