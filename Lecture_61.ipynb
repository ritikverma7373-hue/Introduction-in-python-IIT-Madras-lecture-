# Recursion: find 0 in a list

This is the piece of code to check if is given list has 0 in it, we return false(0)

```python
def check0(l):
# if the list is empty, return false
  if (len(l)==0) :
# if the first element is zero than return 1
    return 0
  if (l[0]==0):
    return 1
  else:
    return check0(l[1:len(l)])
  # example if you have a 10 vessel you just clean 1 vessel
  # then after that you will have 9 vessel so you will go to the 9 vessel.
  # as like you have big list after searching 1 element and
  # and go to small...list ok.
ans=check0([1,2,3,0,4,5,107,2])
print(ans)
```
