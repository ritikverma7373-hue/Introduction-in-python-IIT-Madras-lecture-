# Binary search impletations

```python

def obivious_search(l,k) :
    for x in l :
      if x==k:
        return 1
    return 0
def binary_search(l,k):
  begin=0
  end=len(l)-1
  while (end-begin>1):
    mid=begin+end//2
    if (l[mid]==k):
      return 1
    if (l[mid]>k):
        end=mid-1
    if (l[mid]<k):
        begin=mid+1
    if (l[begin]==k) or (l[end]==k):
       return 1
    else:
      return 0
import time
L=[12,15,1001,1024,2016,2021]
a=time.time();print(obivious_search(L,1024));b=time.time();print(b-a)
a=time.time();print(binary_search(L,11));b=time.time();print(b-a)
l=list (range(1000*1000*100))
a=time.time();print(binary_search(l,12345));b=time.time();print(b-a)
print(len(l))
a=time.time();print(obivious_search(l,-1));b=time.time();print(b-a)
a=time.time();print(binary_search(l,-1));b=time.time();print(b-a)
k=l[len(l)-1]
a=time.time();print(binary_search(l,k));b=time.time();print(b-a)
a=time.time();print(obivious_search(l,k));b=time.time();print(b-a)
a=time.time();print(binary_search(l,-1));b=time.time();print(b-a)
a=time.time();print(obivious_search(l,10));b=time.time();print(b-a)
l=list(range(1000*1000*1000))
a=time.time();print(binary_search(l,-1));b=time.time();print(b-a)# The binary search time taken is 0.0002 sec
a=time.time();print(obivious_search(l,-1));b=time.time();print(b-a)
# The obivious search time taken is 33.78 sec this google colab is crased so you will use another IDE

```
