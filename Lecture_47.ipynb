# Sorting using Functions

We sorted a list of numbers based ona technique called the obivious sort.

```python
# find out the minimum most element in the list l
# append that to a new list x.
# remove the minimum from the orignal list l.
def obivious_sort(l):
 x=[]
 while (len(l)>0) :
    mini=l[0]
    for i in range (len(l)):
     if l[i]<mini:
       mini=l[i]
    x.append(mini)
    l.remove(mini)
 return x
l=[9,8,7,6,5,2,4,3,1]
print(obivious_sort(l))

```
